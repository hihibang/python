# DataFrame Index

## Index란?
- Pandas의 Index는 데이터프레임과 시리즈의 각 행(row)를 고유하게 식별하는 역할을 수행함
- 관계형 데이베이스의 PK와 사한 개념
- 특징
    - 데이터 식별 역할 수행
    - 숫자, 문자열도 가능
    - 생성 후 직접 수정 불가
    - reset_index() 함수를 이용해서 재생성 가능

## 2. Index 기반 연산 방식
- 🌟판다스는 인덱스 이름을 기준으로 데이터 맞춘 후 연산함

### (2) 기본 Series 생성 및 연산

```python
# Series 생성
s1 = pd.Series([10,20,30])
s2 = pd.Series([40,50,60])
print(s1 + s2)
print(s1 - s2)
print(s1 * s2)
print(s1 / s2)
```

```python
#인덱스 지정
s1 = pd.Series([10,20,30], index=['a', 'b', 'c'])
s2 = pd.Series([40,50,60], index=['b', 'c', 'd'])
```

```python
print(s1 + s2)
print(s1 - s2)
print(s1 * s2)
print(s1 / s2)
```

### (2) 데이터 프레임 연산
- 🌟 동일한 인덱스, 동일한 컬럼명 기준으로 연산이 진행됨

```python
df1 = pd.DataFrame(
    {
    'col1': [10, 20, 30],
    'col2': [40, 50, 60]
    }, index = ['a', 'b', 'c']
)
df2 = pd.DataFrame(
    {
    'col1': [10, 20, 30],
    'col3': [40, 50, 60]
    }, index = ['a', 'b', 'c']
)
```

```python
print(df1)
print('-'*30)
print(df2)
print('-'*30)
df1 + df2 # 인덱스 컬럼명이 일치해야 연산이 수행됨
```

```python
# df1객체에 'col3' 컬럼 추가
df1['col3'] = [20, 30, 40]
df1 + df2
```

## 3. reset_index()
- 실행 시 기존 index는 첫번째 컬럼으로 이동하고, 새로운 숫자의 index가 생성됨

```python
# 데이터셋 확인
titanic_df.head()
```

```python
# 인덱스 재생성
td_reset_df = titanic_df.reset_index(inplace=False)
```

```python
td_reset_df2 = td_reset_df.reset_index(inplace=False)
td_reset_df2
```

```python
td_reset_df2.drop('level_0', axis=1, inplace=True)
```

```python
td_reset_df2.head()
```

```python
td_reset_df2.drop([0, 1, 2], inplace=True)
td_reset_df2.head()
```

```python
# td_reset_df2의 인덱스 재생성, 원본반영
td_reset_df2.reset_index(inplace = True)
td_reset_df2.head()
```

```python
td_reset_df2.drop(['level_0', 'index'], axis=1, inplace=True)
td_reset_df2.head()
```

## 4. rename() - 컬럼명, 인덱스명 변경
- 문법
    DF.rename(columns={'before':'after'})

```python
titanic_df['Age_0'] = 0
titanic_df.head(3)
```

```python
titanic_df['Age_0'] = 1
titanic_df.head(3)
```

```python
titanic_df.rename(columns={'Age_0':'Age_1'}, inplace=True)
titanic_df.head(3)
```

```python
# 인덱스 0, 2, 4 > 'a', 'b', 'c'
td_rename = titanic_df.rename(index={0:'a',2:'b', 4:'c'})
td_rename.head()
```

```python
td_rename_index = td_rename.index
td_rename_index # 문자와 숫자가 섞여있음
```
