# DataFrame sort agg groupby

```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

```python
titanic_df = pd.read_csv('data/titanic/train.csv')
titanic_df
```

## 1. DataFrame 정렬(sort)

### (1) sort_values()
- DataFrame을 특정 컬럼 기준으로 정렬할 때 사용
- 문법

  DF.sort_values(by='컬럼', ascending=True)
  - ascending=T 오름차순
  - ascending=F 내림차순

```python
td_sort_name = titanic_df.sort_values(by= 'Name')
td_sort_name.reset_index(inplace=True)
td_sort_name.drop(columns='index', axis=1, inplace=True)
td_sort_name.head()
```

```python
td_sort_pclass = titanic_df.sort_values(by='Pclass', ascending=False)
td_sort_pclass.reset_index(drop=True, inplace=True) # drop=True 기존 인덱스는 삭제
td_sort_pclass.head()
```

```python
titanic_df.sort_values(by=['Name', 'Pclass']).head(3)
```

## 2. 집계 함수(Aggregation)

### (1) count()
- 결측치 제외
- 실제 데이터 갯수 반환

```python
titanic_df.count()
```

### (2) shape
- 전체 데이터 수를 확인하는 작업에서는 count() 함수보다 shape을 사용하는 것이 유리

```python
titanic_df.shape
```

### (3) mean()
- 평균

```python
titanic_df['Age'].mean()
```

```python
print(titanic_df['Fare'].mean().round())
```

### (4) sum()
- 합계 집계 함수

```python
titanic_df.head()
```

```python
titanic_df[['Age', 'Fare']].sum().round(2)
```

### (5) min()/ max()
- 최소/최대값 집계 함수

```python
print(titanic_df[['Age', 'Fare']].min())
print('-'*40)
print(titanic_df[['Age', 'Fare']].max())
```

## 3. Group By
- 특정 컬럼 기준으로 데이터를 그룹화할 때 사용
- 문법
  DF.GroupBy('컬럼리스트')

```python
# 객실 등급별 이원수 조회
titanic_df.groupby('Pclass').count()
```

```python
titanic_df.groupby('Embarked').count()
```

## 4. agg()
- 집계 함수
- 그룹별 서로 다른 집계 함수 적용

```python
titanic_df.agg(
    {
        'Pclass' : 'count',
        'Fare' : 'sum'
    }
)
```

```python
titanic_df.groupby('Embarked').agg(
    {
        'Pclass' : 'count',
        'Fare' : 'sum'
    }
)
```

```python
# 객실 등급별 그룹, age 최대 fare 평균
titanic_df.groupby('Pclass').agg(
    {
        'Age':'max',
        'Fare':'mean'
    }
)
```

```python
# 여러 컬럼에 agg() 함수 적용
titanic_df[['Age', 'Fare']].agg('max')
```

```python
titanic_df.agg(
    {
        'Age': 'mean',
        'Fare': 'sum',
        'Pclass': 'count',
    }
)
```
