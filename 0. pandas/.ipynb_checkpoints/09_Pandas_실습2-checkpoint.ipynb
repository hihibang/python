```python
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import platform
import pandas as pd

if platform.system() == 'Windows':
    matplotlib.rc('font', family='Malgun Gothic')
elif platform.system() == 'Darwin':
    matplotlib.rc('font', family='AppleGothic')
else:
    matplotlib.rc('font', family='DejaVu Sans')

plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 100

print('NumPy 버전:', np.__version__)
print('환경 설정 완료')
```

```python
# 데이터 로드
df1 = pd.read_csv('data/서울특별시 공공자전거 일별 대여건수_25.7-12.csv')
df1.head()
```

```python
# 데이터 로드
df2 = pd.read_excel('data/대여소별 대여내역(2017년).xlsx')
df2.head()
```

```python
# df1의 일자별 대여건수 top10
df1.shape
df1['대여일자'].value_counts()
df1['대여일자'].dtypes
```

```python
# '대여일자' 데이터 타입 변환, to_datetime()
df1['대여일자'] = pd.to_datetime(df1['대여일자'])
df1.dtypes
df1['Year'] = df1['대여일자'].dt.year
df1['Month'] = df1['대여일자'].dt.month
df1['Day'] = df1['대여일자'].dt.day
df1['Month'].value_counts()
```

```python
# df1의 2025년도 월별 대여건수 
df1_rentals = df1.groupby('Month')['Month'].sum()
df1_rentals
df1_rentals.index
#df1_rentals.values
```

```python
# 막대 그래프
plt.figure(figsize=(20, 5))
plt.bar(df1_rentals.index, df1_rentals.values, color='coral')
plt.title('2025년도 월별 대여건수')
plt.xlabel('월')x
plt.ylabel('대여건수')
# plt.gca().invert_xaxis()
# plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# df2의 대여소별 대여건수 top5
df2.head()
#df2['대여소 명'].value_counts()
```

```python
# 요일별 대여건수 그래프 시각화
df1.dtypes
df1['대여일자'].dt.day_name()
```

```python
# 0: 월 1: 화 .... 
df1['대여일자'].dt.weekday
```

```python
# 요일을 매핑하는 딕셔너리 객체 생성
week_dict = {
    0: '월',
    1: '화',
    2: '수',
    3: '목',
    4: '금',
    5: '토',
    6: '일'
}
df1['weekday'] = df1['대여일자'].dt.weekday.map(week_dict)
```

```python
df1['weekday'].value_counts()
order = ['월','화','수','목','금','토','일']
result = df1.groupby('weekday')['대여건수'].sum().reindex(order)
result
```

```python
# 막대 그래프
plt.figure(figsize=(20, 7))
plt.bar(result.index, result.values, color='pink')
plt.title('2025년도 요일별 대여건수')
plt.xlabel('월')
plt.ylabel('대여건수')
# plt.gca().invert_xaxis()
# plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df2.head()
df2.shape
df2.info()
```

```python
df2.rename(columns={'대여소 그룹':'그룹'})
df2.columns = ['그룹','대여소','일자','건수']
df2
```

```python
df2.groupby('대여소')['건수'].sum().sort_values(ascending = False).head()
```

```python
# 그룹별 총 대여건수 탑 10
results = df2.groupby('그룹')['건수'].sum().sort_values(ascending=False).head(10)
```

```python
# 월별 전체 대여건수
df2['월'] = df2['일자'].astype(str).str[5:7]
results = df2.groupby('월')['건수'].sum().sort_values(ascending=False).head(10)
results
```

```python
# 대여소별 평균 대여건수 15
results = df2.groupby('대여소')['건수'].mean().sort_values(ascending=False).head(15)
results
```

```python
# 막대 그래프
plt.figure(figsize=(20, 7))
plt.bar(results.index, results.values, color='orange')
plt.title('막대 그래프')
plt.xlabel('X_label')
plt.ylabel('Y_label')
# plt.gca().invert_xaxis()
# plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
