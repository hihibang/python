# 이변수 데이터 분석

```python
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import platform
import pandas as pd

if platform.system() == 'Windows':
    matplotlib.rc('font', family='Malgun Gothic')
elif platform.system() == 'Darwin':
    matplotlib.rc('font', family='AppleGothic')
else:
    matplotlib.rc('font', family='DejaVu Sans')

plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 100

print('NumPy 버전:', np.__version__)
print('환경 설정 완료')
```

```python
kobis = pd.read_csv('data/kobis.csv')
kobis.head()
kobis.info()
```

```python
kobis['일일매출액'].value_counts() # 수치형 가능성 ⬆️
```

```python
# 무게와 연비의 상관계수 구하기 df['무게'].corr(df['연비')
```

```python
# 일일매출액과 일일관객수의 상관계수
r= kobis['일일매출액'].corr(kobis['일일관객수'])
print(f'r(상관계수): {r:3f}') # 이상치 데이터에 영향을 받음
```

### iris 데이터셋의 꽃잎길이(PetalLength)와 꽃잎넓이(PetalWidth)의 상관관계 도출

```python
iris = pd.read_csv('data//iris.csv')
iris['Species']
```

```python
iris.info()
```

```python
# 통계지표 확인
iris.describe()
```

```python
# 상관계수 도출
r= iris['PetalLengthCm'].corr(iris['PetalWidthCm'])
print(f'꽃잎의 길이와 넓이의 상관계수: {r:3f}')
```

```python
# 산점도 그래프 생성
iris.plot(
    x='PetalLengthCm',
    y='PetalWidthCm',
    kind='scatter'
)
```

```python
colors = {
    'Iris-setosa':'red',
    'Iris-versicolor':'green',
    'Iris-virginica':'blue'
}
colors = [colors[sp] for sp in iris.Species]
colors
```

```python
iris.plot.scatter(
    x='PetalLengthCm',
    y='PetalWidthCm',
    c=colors
)
plt.show()
```

### 타이타닉 데이터셋의 나이와 요금의 상관 관계 도출


```python
titanic = pd.read_csv('data/titanic/train.csv')
titanic.info()
titanic.head()
titanic['Embarked'].value_counts()
```

```python
titanic['Age'].fillna(titanic['Age'].mean())
titanic.describe()
```

```python
# 상관계수 도출
r= titanic['Age'].corr(titanic['Fare'])
print(f'상관계수: {r:2f}')
```

```python
colors = {
    'S': 'green',
    'C':'blue',
    'Q':'yellow'
}
colors = [colors.get(sp, 'gray') for sp in titanic.Embarked]
```

```python
# 상관 그래프 도출
titanic.plot.scatter(
    x='Age',
    y='Fare',
    c=colors
)
plt.show()
```

```python
# 상관 그래프 도출
titanic['Age'].plot.box()
plt.show()
```
