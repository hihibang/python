# DataFrame과 리스트 딕셔너리 ndarraty 상호 변환

### 1. 1차원 데이터를 DataFrame으로 변환

```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

```python
# 1차원 리스트 생성
col_name1 = ['col1']
list1 = [1,2,3]
print(list1, type(list1))
```

```python
# ndarray로 변환
arr1= np.array(list1)
print(arr1, type(arr1))
```

```python
# list1, arr1을 DataFrame으로 생성
df1_1 = pd.DataFrame(list1, columns=col_name1, index=['a', 'b', 'c'])
df1_2 = pd.DataFrame(arr1, columns=col_name1)
```

```python
print(df1_1)
print(df1_2)
```

### 2. 2차원 데이터를 DataFrame으로 변환

```python
# 컬럼명 생성
col_names = ['col1', 'col2', 'col3']
col_names
```

```python
# 2차원 리스트 생성
list2 = [[1,2,3], [4,5,6]]
list2
```

```python
#mdarray로 변환
arr2 = np.array(list2)
arr2
```

```python
# list2, arr2를 DataFrame으로 변환
df2_1 = pd.DataFrame(list2, columns = col_names)
df2_2 = pd.DataFrame(arr2, columns = col_names)
```

```python
print(df2_1)
print(df2_2)
```

### 3. 딕셔너리를 DataFrame으로 변환

```python
# 딕셔너리 생성
dict1 = {
    'col1': [1,2,3],
    'col2': [4,5,6],
    'col3': [7,8,9]
}
dict1
```

```python
df_dict1 = pd.DataFrame(dict1)
df_dict1
# 시리즈가 여러개 묶인 것 = DataFrame
```

### 4. DataFrame을 ndarray, list, dictionary로 변환

```python
# df_dict1 데이터프레임을 ndarray로 변환
# arr_dict1 = np.array(df_dict1)
arr_dict1 = df_dict1.values
arr_dict1
```

```python
# df_dict1 데이터프레임을 list로 변환
list3 = df_dict1.values.tolist() # values까지는 array로 변환
list3
```

```python
# df_dict1 데이터프레임을 dictionary로 변환
dict3 = df_dict1.to_dict()
dict3
```

### 5. shape 속성 정리
- python: List : ❌ AttributeError
- python: Dictionary  : ❌ AttributeError
- python: Ndarray : 🅾️ 사용가능
- pandas: DataGFrame : 🅾️ 사용가능
- pandasL Series : 🅾️ 사용가능

```python
# list 생성
list1
print(list1.shape) # ❌ AttributeError: type object 'list' has no attribute 'shape'
```

```python
# dict 생성
dict1
print(dict1.shape) # ❌ AttributeError: 'dict' object has no attribute 'shape'
```

```python
df_dict1
print(df_dict1.shape) # (3, 3)
```

### 6. DataFrame 컬럼 데이터 생성 및 구성

```python
import pandas as pd
```

```python
titanic_df = pd.read_csv('data/titanic/train.csv')
titanic_df.head(3)
```

```python
# 기억하면 좋은 잘쓰는 것 🌟
print(titanic_df.head()) # 상위 5개
print('*'*100)
print(titanic_df.tail()) # 하위 5개
print('*'*100)
print(titanic_df.info()) # 기본 정보
print('*'*100)
print(titanic_df.describe()) # 수치 정보
```

```python
# Age_0 컬럼 추가
# titanic_df.info()
titanic_df['Age_0'] = 1 # 컬럼 🅾️ 값 수정, 컬럼 ❌ 컬럼 추가
titanic_df.info()
```

```python
# Age_by_10 컬럼 추가, Age*10
titanic_df['Age_by_10'] = titanic_df['Age']*10
titanic_df
```

```python
# 일부 컬럼 조회
titanic_df['Age']
```

```python
# 여러개의 컬럼 조회
titanic_df[['Age', 'Age_0', 'Age_by_10']].head()
```

```python
# Family_no 컬럼 추가 값 sibSp + Parch +1
titanic_df['Family_no'] = titanic_df['SibSp'] + titanic_df['Parch'] +1
titanic_df.info()
titanic_df.head()
```

### 7. DataFrame 데이터 삭제

```python
# Age_0 컬럼 삭제
td_drop = titanic_df.drop('Age_0', axis = 1) # 기본값이 axis=0이 라서 행에서 컬럼을 찾음 KeyError: "['Age_0'] not found in axis"
```

```python
titanic_df.head() # Age_0 🅾️
td_drop.head() # Age_0 ❌
```

```python
# 원본 삭제 후 반영
titanic_df.drop(['Age_by_10', 'Family_no'], axis=1, inplace=True)
titanic_df.head() # Age_0 ❌
```

```python
titanic_df.drop([0, 1, 2], inplace=True) # 인덱스 행 기준 0번째 삭제
```
