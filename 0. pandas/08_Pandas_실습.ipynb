## 6. 연산

```python
df['승차총승객수_10'] = df['승차총승객수'] + 10
df
```

```python
df['승하차총승객수'] = df['승차총승객수'] - df['하차총승객수']
df
```

```python
print(df['승하차총승객수'].mean())
print(df['승하차총승객수'].median())
print(df['승하차총승객수'].agg(['mean', 'median']))
```

```python
df['승하차총승객수합계'] = df['승차총승객수'] + df['하차총승객수']
df
```

```python
print(df['승하차총승객수합계'].agg(['mean', 'median']))
```

```python
# 타이타닉 데이터셋 로드
td = pd.read_csv('data/titanic/train.csv')
```

```python
td.isnull().sum()
```

```python
td.drop(columns = 'Cabin', inplace = True)
```

```python
td.isnull().sum()
```

```python
# Nan 값을 이전 값으로 대체
td['Embarked'] = td['Embarked'].ffill()
```
