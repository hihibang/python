```python
import pandas as pd
import matplotlib.pyplot as plt
fruits = pd.read_csv('data/fruits.csv')
fruits.info()
```

```python
apple = fruits[:35]
```

```python
plt.scatter(
    apple.length,
    apple.weight
)
plt.show()
```

```python
berry = fruits[35:]
berry.info()
```

```python
plt.scatter(
    berry.length,
    berry.weight
)
plt.show()
```

```python
plt.scatter(
    fruits.length,
    fruits.weight
)
plt.show()
```

```python
plt.scatter(
    apple.length,
    apple.weight
)
plt.scatter(
    berry.length,
    berry.weight
)
plt.show()
```

### 머신러닝을 위한 데이터 준비

```python
fruits_data = fruits[['length', 'weight']]
fruits_target = fruits['name']
```

## 최근접이웃 분류 알고리즘

```python
from sklearn.neighbors import KNeighborsClassifier
kn = KNeighborsClassifier() # default 5, 하이퍼파라미터 조정: n_neighbors=3
```

```python
kn.fit(fruits_data, fruits_target)
```

```python
kn.score(fruits_data, fruits_target)
```
