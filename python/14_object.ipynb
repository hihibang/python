## (1) 클래스 정의
```
class 클래스명():
    변수1
    변수2

    def 함수1(self, [인자1, 인자2, 인자3])
        코드
    def 함수2(self, [인자1, 인자2, 인자3])
        코드

# 선언
변수 = 클래스명(인자 ...)
# 호출
변수.함수()
```

````python
class Animal():
    className = "Animal"

    def __init__(self, name, emoji):
        self.name = name
        self.emoji = emoji

    def display(self):
        print(f'name: {self.name}, emoji: {self.emoji}')
````

````python
dog = Animal("해피", "🐶")
dog.display()
````

````python
class Bycycle():
    className = "Bycycle"

    def __init__(self, size, color):
        self.size = size
        self.color = color

    def info(self):
        print(self.size, self.color)

    def move(self, speed):
        print(f'자전거가 {speed}로 움직입니다')
````

````python
mycycle = Bycycle(50, "Red")
mycycle.info()
````

````python
mycycle.move(30)
````

## (2) 클래스 상속

````python
# Animal 클래스의 자식 Dog 클래스 정의
class Dog(Animal): 
    className = "Dog"

    def __init__(self, name, emoji, color):
        super().__init__(name, emoji) # 부모에서 가져와서
        self.color = color

    def display(self):
        print(self.name, self.emoji, self.color)

    def setName(self, name):
        self.name = name

    def getName(self):
        return self.name
````

````python
happy = Dog("해피", "🐶", "갈색")
happy.display()
print(Dog.className)
happy.setName("초코")
happy.display()
happy.getName()
````

````python
class Cat(Animal):
    def __init__(self, name, emoji, color, age, ctype):
        super().__init__(name, emoji)
        self.color = color
        self.age = age
        self.ctype = ctype

    def display(self):
        print(self.name, self.emoji, self.color, self.age, self.ctype)
    
    def setName(self, name):
        self.name = name
    def setEmoji(self, emoji):
        self.emoji = emoji
    def setColor(self, color):
        self.color = color
    def setAge(self, age):
        self.age = age
    def setCtype(self, ctype):
        self.ctype = ctype
        
    def getName(self):
        return self.name
    def getEmoji(self):
        return self.emoji
    def getColor(self):
        return self.color
    def getAge(self):
        return self.age
    def getCtype(self):
        return self.ctype
````

````python
jump = Cat("jump", '🐱', '갈색', 9, 'home')
jump.setAge(3)
jump.display()
print(jump.getAge())
````
