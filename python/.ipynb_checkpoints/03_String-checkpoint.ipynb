# 문자열

- 파이썬에서는 따옴표('',"")로 둘러싼 문자 집합을 의미

# 1. 문자열 생성
- 시작과 끝에 같은 따옴표 사용

```python
print("Hello")
print('Hello')
```

```python
str1 = "Hello"
str2 = 'Hello'
print(str1, str2)
```

```python
print(type(str1), type(str2))
```

```python
str3 = "제 이름은 '홍길동' 입니다"
print(str3)
```

```python
# 문자열을 여러 줄 입력하는 방법
str4 = "파이썬을 배우는 중입니다"
str5 = """ 
파이썬을
배우는
중입니다 """
print(str4, str5)
str5
```

# 2. 문자열 다루기
- 더하기 연산자 + 와 곱하기 연산자 *

```python
a = "Hello~ "
b = "Python "
print(a+b)
print(a*3)
print(b*3)
```
