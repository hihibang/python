```python
# a라는 변수의 10 입력
a = 10 # 10이라는 객체를 만들고 a는 10 객체가 위치하는 주소값
print('a= ', a)
```

```python
ABC = 200
ABC
```

```python
# 여러개의 변수 선언 가능
# x, y, z = 100 # ❌
x, y, z = 100, 200, 300 # 🅾️
print(x, y, z)
```
