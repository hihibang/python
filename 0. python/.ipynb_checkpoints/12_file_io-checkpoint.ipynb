# file 열기(open())

- 내장함수인 오픈 함수를 이용하여 파일을 연 다음 Read, Write가 가능함
- 형식 file = open(파일주소)
- 속성:
- 읽기(r)
- 쓰기(w)
- 추가(a): append, 같은 이름의 파일 없으면 w로 추가 생성
- b: 바이너리 파일 모드로 열기
- t: 텍스트 파일 모드로 열기

## 1. 파일 쓰기

- 파이르쓰기를 하려면 open() 함수의 쓰기 모드로 열어야 함
- 파일 쓰기를 종료하면, 반드시 파일을 닫아야 함

```python
#myfile.txt 파일을 생성
f = open('myfile.txt', 'w')
f.write('Hello MyFile')
f.close()
```

```python
#myfile.ko.txt 파일을 생성
k = open('myfile.ko.txt', 'w', encoding='cp949')
k.write('안녕 마이파일')
k.close()
```

## 2. 파일읽기

```python
f = open('myfile.txt', 'r')
f_txt = f.read()
f.close() # 파일 객체 종료

f_txt
```

```python
k = open('myfile.ko.txt', 'r', encoding='cp949')
k_txt = k.read()
k.close() # 파일 객체 종료

k_txt
```

## 3. 파일에 문자열 한 줄씩 쓰기

```python
f=open('myfile_multi.txt', 'w')
for i in range(1, 6):
    str = f'number= {i}\n'
    f.write(str)
f.close()
```

## 4. 파일에 문자열 한 줄씩 쓰기

```python
f = open('myfile_multi.txt', 'r')
line = f.readline()
while line:
    print(line)
    line  = f.readline()
f.close()
```

```python
# readlines() 파일 전체를 한 줄씩 읽어 리스트 객체로 변환

f = open('myfile_multi.txt', 'r')
fl = f.readlines()
f.close()

fl
```

```python
# 추가모드 실행
f = open('myfile.txt', 'a')
f.write('open')
```
