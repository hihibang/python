### (1) 인자도 반환값도 없는 함수

````python
# 함수 생성
def f1():
    print('f funtion~1')
````

````python
print(dir())
type(f1)
````

````python
# 함수 호출
f1()
````

### (2) 인자값은 있는 함수

````python
# 함수 생성
def f2(name, age):
    print(f'Name: {name} \n Age: {age}')
````

````python
print(dir())
type(f2)
````

````python
# 함수 호출
f2('kim', 20)
````

### (3) 인자값과 반환값이 있는 함수

````python
# 함수 생성
def f3(x, y):
    return x+y
````

````python
print(dir())
type(f3)
````

````python
# 함수 호출
f3(2, 3)
````

구구단 함수 생성 및 호출

````python
# 함수 생성
def gugudan(n):
    for i in range(1, 10):
        print(f'{n} X {i} = {n*i}')
````

````python
gugudan(2)
````

````python
def gugudan_multi(start, end):
        for n in range(start, end+1):
            for i in range(1, 10):
                print(f'{n} X {i} = {n*i}')
````

````python
gugudan_multi(2, 10)
````

간단한 사칙연산 계산기

````python
def cacl(n1, n2, cacl):
    if cacl == '+':
        return n1+n2
    elif cacl == '-':
        return n1-n2
    elif cacl == 'x':
        return n1*n2
    elif cacl == '/':
        return n1/n2
    else:
        print('Error')

# def cacl(n1, n2, cacl):
    
#     return n1  n2
````

````python
cacl(2, 3, '-')
````
