구구단 함수 생성 및 호출

````python
# 함수 생성
def gugudan(n):
    for i in range(1, 10):
        print(f'{n} X {i} = {n*i}')
````

````python
gugudan(2)
````

````python
def gugudan_multi(start, end):
        for n in range(start, end+1):
            for i in range(1, 10):
                print(f'{n} X {i} = {n*i}')
````

````python
gugudan_multi(2, 10)
````

간단한 사칙연산 계산기

````python
def cacl(n1, n2, cacl):
    if cacl == '+':
        return n1+n2
    elif cacl == '-':
        return n1-n2
    elif cacl == 'x':
        return n1*n2
    elif cacl == '/':
        return n1/n2
    else:
        print('Error')

# def cacl(n1, n2, cacl):
    
#     return n1  n2
````

````python
cacl(2, 3, '-')
````

## 3. 변수의 Scope(유효 범위)
- 전역변수(global): 전역변수는 함수의 전체에서 호출됨, 된 지역변수로 같은 이름의 변수 선언시 **지역변수 > 전역변수 호출이 우선**
- 지역변수(local): 함수 안에서 정의한 변수는 함수 안에서만 호출됨
- global: 함수안에서 전역변수를 호출해서 사용하는 키워드

````python
a = 5 #전역변수

def f1():
    a = 1
    print(a)

def f2():
    print(a)

def f3():
    global a # 전역변수 a를 함수 안에서 정의
    a = 10 # 지역이 아니라 전역변수 정의
    print(a)
def f4():
    print(a)
````

````python
f1() #지역변수인 a = 1이 출력
f2() #전역변수인 a = 5이 출력
f3() #지역변수인 a = 10이 출력
f4() #전역변수인 수정된 a = 10이 출력
````

## 4. 개수가 정해지지 않은 인자
- *args
- **kwargs: 딕셔너리{키:값}로 값을 전달
- 여러 개의 인자를 함수로 매핑하는 경우 **튜플로 저장**
- *args 일반 매개변수보다 항상 **뒤에** 위치


````python
def func(a, b, *args): # argment
    print(a, b, args)
func(1,2,3,4,5,6,7) # parameter 매개인자
func(1, 10)
````

````python
def func2(a, b, **kwargs):
    print(a, b, kwargs, type(kwargs))
func2(1,2, x=1,y=2,z=3) # parameter 매개인자
````

````python
def func2(a, b, **kwargs):
    print(a, b, kwargs, type(kwargs))
    for k in kwargs: # kwargs.keys() => 리스트, 알아서 키 값만 출력되게 만들어 줌
        print(k)
func2(1,2, x=1,y=2,z=3) # parameter 매개인자
````

## 5. 람다
- 무기명 함수를 선언하는 방법
- 한줄로 함수를 표현
- 형식:
1    (lambda 인자: 실행문) (호출)
2    변수 = lambda 인자: 실행문
     변수(인자)

````python
    (lambda x: x+1) (3)    
````

````python
square = lambda x: x**2
print(square(2))
square2 = lambda x, y: x**2 + y
square2(2, 1)
````

### filter 함수 => 조건에 맞는 데이터를 추출하는 함수
- filter(함수, 데이터)

````python
#리스트에 지정된 숫자 둥 짝수를 출력
numbers = [1,2,3,4,5]
result = list(filter(lambda n: n%2 == 0 , numbers))
````

````python
for i in result:
    print(i)
````

````python
# numbers에 *2 적용하여 새로운 리스트 생성
list(map(lambda n : n*2, numbers))
````
