# 컨테이너 타입이란?

- 컨테이너 타입은 여러 개의 값을 다루는 자료형입니다. **list, tuple, set, dictionary**가 있습니다.
- **리스트(list)** : 데이터를 연속적으로 관리, 값을 바꿀 수 있음, 예:['가', '나', '다', '라']

- **튜플(tuple)** : 데이터를 연속적으로 관리, 값을 바꿀 수 없음, 예:('가', '나', '다', '라')

- **세트(set)** : 집합과 같은 속성, 예:{'가', '나', '다', '라'}

- **딕셔너리(dictionary)** : key-value 쌍, 예: {'가':123, '나':456, '다':789}

## List(리스트)

여러 개의 데이터를 묶어서 처리할 때 사용

### 1. List 생성

- 대괄호 [] 사용, 각각의 데이터는 콤마로 구분
- [] 입력 시 빈 리스트 생성
- List의 데이터는 다양하게 저장 가능

```python
# 1반 학생의 시험 성적을 저장 - 국어, 영어, 수학 기준 90, 80, 100
score = [90, 80, 100]
score
```

```python
# 1반 학생이름 저장
names= ["홍씨", "박씨", "김씨"]
type(names)
```

### List에서 아이템 지정 방법
- 리스트명[인덱스]

#### 아이템 조회

```python
# 국어 영어 수학 중 국어 점수 조회
score[0]
```

```python
# 리스트의 크기 출력
len(score) # score.length
```

```python
score[-1] # -1 인덱스의 마지막 아이템 선택
```

```python
empty_list = []
empty_list
```

#### 아이템 변경
- 리스트명[인덱스] = 뉴 아이템

```python
print(names[0])
names[0] = "파이썬"
print(names[0])
```

```python
# 학생명 국어 영어 수학 데이터 저장
class1 = ["강씨", 90, 90, 100, True, "김씨", 50, 30, 40]
class1
```

### 3. List 연산
- 더하기: 두 개의 리스트 연결
- 곱하기: 곱한 수 만큼 반복
- 범위지정(클론, :): 리스트[시작 : 끝] 리스트[시작 : 끝 : 간격]
- 아이템 제거: del 리스트[인덱스]
- 리스트 조회: 아이템 in 리스트객체 

```python
# 더하기
list1 = [1,2,3,4]
list2 = [2,3,4,5]
list1 + list2
```

```python
# 곱하기
print(list1 * 2)
print(list2 * 3)
```

### 리스트 범위 정의

```python
# 인덱스의 범위를 정의하여 아이템 조회
list3 = (list1 * 2) + list2
list3[2: 10]
```

```python
print(list3[5 :len(list3)])
```

```python
print(list3[2: ])
print(list3[: 4])
print(list3[:])
print(list3[:: 2])
```

```python
del list3[:]
list3
```

#### 아이템 존재 여부 확인

```python
list4 = (list2 * 2) + list1
list4
9 in list4
```

## 4. 리스트의 메소드
- 리스트.메소드()
- (1) append 리스트의 마지막에 새로운 항목 추가
- (2) insert 특정 위치에 아이템 추가 insert(인덱스, 추가할 항목)
- (3) extend 여러개의 항목을 마지막에 추가 extend([추가할 항목 1, 2 ...])
- (4) remove 항목 삭제하기 remove(제거하려는 항목)
- (5) pop 마지막 아이템 삭제
- (6) index 아이템의 현재 인덱스 확인 index(확인할 항목)
- (7) count 아이템의 전체 갯수 확인 count(확인할 항목)
- (8) sort 오름차순 정렬
- (9) revers 내림차순 정렬

```python
# append
fruits = ["Apple", "Banana"]
fruits
```

```python
fruits.append("orange")
fruits
```

```python
# insert
fruits.insert(0, "tomato")
fruits
```

```python
# extend
fruits.extend(["strowbarry", "bae"])
fruits
```

```python
# remove 다중 아이템 삭제 불가
fruits.remove("strowbarry")
fruits
```

```python
# pop
fruits.pop()
fruits
```

```python
# index
fruits.index("Apple")
```

```python
# count
list3 = (list2*2) + list4 + list1
print(list3)
list3.count(2)
```

```python
# sort
list3.sort()
list3
```

```python
# reverse
list3.reverse()
list3
```
