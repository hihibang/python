# 배열 연산
- 데이터셋의 쉐입(배열)이 같으면 연산이 가능)

```python
import numpy as np
```

```python
arr1= np.array([1,2,3])
arr2= np.array([4,5,6])
```

```python
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
```

```python
arr1= np.array([[1, 2], [3, 4]])
arr2= np.array([[5, 6], [7, 8]])
print(arr1.shape)
print(arr2.shape) # 둘의 쉐입이 같아야 연산이 가능하다
```

```python
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
```

# 정렬 - sort()와 argsort()

## (1) sort
- 원본유지 🅾️: numpy.sort()
- 원본유지 ❌: numpy.ndarray.sort()

```python
org_arr = np.array([2,1,3,6,9])
org_arr
```

```python
# 오름차순 정렬 후 배열 반환
sort_arr = np.sort(org_arr)
print(sort_arr)
print(org_arr, sort_arr)
```

```python
arr = np.array([[5, 2], [10, 8]])
```

```python
sort_arr = np.sort(arr, axis=0)
sort_arr #정렬이 안댐
```

```python
sort_arr2 = np.sort(arr, axis=1)
sort_arr2 #작은 것에서 큰 것으로 정렬
```

# argsort()
- 배열을 정렬 후 반환

```python
names = np.array(['홍', '김', '최', '산', '강'])
scores = np.array([90, 87, 56, 94, 40])
```

```python
# 성젹순으로 names 배열을 오름차순 정렬 [::1]
print(names[scores.argsort()])
print(names[np.argsort(scores)])
```

```python
# 성적순으로 names 배열을 내림차순으로 정렬
index = np.argsort(scores)[::-1]
print(names[index])
```

# 집계 함수

```python
arr = np.arange(1, 10)
arr
```

```python
print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(np.std(arr)) # 표준편차
print(np.var(arr)) # 분산
print(np.median(arr)) # 중앙값
```

```python
arr = np.array([[1,2,3], [4,5,6]])
print(np.sum(arr))
print(np.sum(arr, axis=1)) # 열방향 합계
print(np.sum(arr, axis=0)) # 행방향 합계
```

# 선형 대수 연산

## (1) 행렬 내적
- 행렬 내적의 곱셈은 **dot** 함수 사용
- 예) A(2,2) * B(3,2) = C(2,2)
- 규칙: A의 열 수 == B의 행 수가 동일할때만 내적 연산이 가능함
- np.transpose() 함수를 이용하여 행과 열을 바꿈

```python
A = np.array([[1,2,3], 
             [4,5,6]])
B = np.array([[7, 8], 
             [9, 10], 
             [11,12]])
```

```python
print(np.dot(A, B))
```

```python
A_T = np.transpose(A)
A_T
```

```python
A_T * B
```

# 기타 함수
- np.where(): 조건부 값 출력
- np.uniqe(): 중복 없는 고유한 값
- np.concatenate(): 배열을 결합

```python
arr = np.arange(1, 10)
```

```python
print(np.where(arr>5, 'big', 'small'))
```

```python
# np.uniqe() 중복 없는 고유한 값을 반환
arr = np.array([1, 2, 2, 5, 3, 3, 2, 1, 6, 4, 5, 4, 7])
```

```python
np.unique(arr)
```

```python
values, counts = np.unique(arr, return_counts=True)
```

```python
print(values, counts)
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
c= np.concatenate([a, b])
c
```
