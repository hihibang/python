### Deep Learning
- 머신러닝의 한 종류
- 인공신경망은 인간의 뇌신경망의 신호 체계를 본따서 만든 구조
- 이미지 인식, 음성 인식, 자연어 처리 등의 다양한 분야에서 활용

#### 딥 러닝과 머신 러닝의 차이점
- 가장 큰 차이점은 '특징량 추출'의 유무
- 머신러닝은 사람(분석가)이 이미지 내부에 있는 특징을 지정하는 방식
- 딥러닝에서는 학습데이터에서 기계가 자동으로 특징을 추출
- 딥 러닝은 스스로 학습을 하기 때문에 머신 러닝보다 더 많은 데이터가 필요함
- 즉, 머신러닝 까지는 인간이 하나하나 가르쳐야하지만 딥러닝을 이용하면 기계가 자동으로 특징을 학습할 수 있다.

#### 신경망
- 인간의 신경망을 본따 만든 네트워크 구조를 의미
- 입력층, 은닉층, 츨력층을 지나 처리가 일어나고 최종적인 결과가 출력
- 입력층이 3개라면 Feature가 3개라는 의미
- 입력층 + 출력층 = 신경망 + 은닉층(2개 ⬆️) = 깊은 신경망
- 이러한 신경망을 은닉층의 2개 이상 추가 되면 깊은 신경망이라 하며 이를 활용한 기계학습을 딥러닝이라 부름
- 딥러닝은 다량의 데이터를 입력해서 학습한다.
- 학습이란 유닛의 접속 가중치를 조정(update)하는 것을 의미

#### 최적화
- 대표적으로 경사하강법
- 반복적우로 손실함수에 대한 모델 매개변수의 기울기를 구한 후 그 값의 반대값으로 조절해 나가면 결국에는 최저 매개변수에 도착한다,

#### 역전파
- 옵티마이저는 손실함수의 값을 최적화하기 위해 역전파를 사용해 딥러닝의 모든 매개변수 변경
- 손실함수의 값을 최소화: 정답과 예측값의 차이를 최소화, 에러율을 최저로 줄인다는 의미

#### 옵티마이저
- Batch 경사 하강법
- SGD(Stochastic Gradient Descent)

#### 미니 배치
- 배치 경사법은 느리고, 리소스도 많이 사용, SGD는 확실히 빨리 학습하지만 모델이 최적화 안되는 경우도 많음
- 두 방법의 절충안이 미니 배치
- 전체 데이터를 계산해서 매개변수를 변경하는 대신 정해진 양만큼 계산해서 매개변수를 최적화 하는 방법
- 최근 딥러닝 학습에서는 대다수가 미니배치를 사용

#### 학습률
- Adagrad
- Adam

#### 딥러닝의 과대적합
- 과대적합은 검증시의 예측력 차이가 심한 현상
- 딥러닝 모델은 학습시 매개변수가 많다는 점과 학습 횟수의 제한이 없으며 이 두 특징은 과대적합을 불러 올 수 있다.

#### 드롭아웃
- 드롭아웃은 매개변수 중 일정량을 학습 중간마다 무작위로 사용하지 않는 방법

#### 조기종료
- 무조건 학습 반복 횟수(Epoch)를 높이면 학습시간이 길어져서 학습 데이터만 성능이 좋은 현상이 발생
- 어느 정도 학습률이 좋으면 사용자가 정한 반복횟수 전에 작업을 종료한다
