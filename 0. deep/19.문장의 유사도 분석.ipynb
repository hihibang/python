```python
a = '딥러닝은 매우 재미있는 기술이라 공부할 수 있습니다'
b = '공부하면 재미있는 기술이라 딥러닝을 배우고 있습니다.'
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
sentences = (a, b)

tfid_fVectorizer = TfidfVectorizer()
```

```python
sentences
```

```python
# 문장 백터화 하기
tfid_martix = tfid_fVectorizer.fit_transform(sentences)
tfid_martix
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cos_similar = cosine_similarity(tfid_martix[0:1], tfid_martix[1:2])
cos_similar
```
